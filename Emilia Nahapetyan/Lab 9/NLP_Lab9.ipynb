```python
import logging
logging.basicConfig(
    format='%(asctime)s : %(levelname)s : %(message)s',
    level=logging.INFO
)
```

```python
import os
import re

def clean_text(text):
    text = re.sub(r'\d+', ' ', text)
    text = re.sub(r'[“”"\'«».,!?;:()<>/\\@#$%^&*+=_—-]', ' ', text)
    text = text.replace("եւ", "և")
    text = re.sub(r'\s+', ' ', text).strip()
    return text

corpus_texts = []

root_path = "ilur-news-corpus/ilur-news-corpus/test"

for root, dirs, files in os.walk(root_path):
    for fname in files:
        if fname.endswith(".txt"):
            path = os.path.join(root, fname)
            with open(path, encoding='utf-8') as f:
                text = clean_text(f.read())
                corpus_texts.append(text)
```

```python
print("Loaded documents:", len(corpus_texts))
print("Sample:", corpus_texts[0][:200])
```

```python
with open("armenian_corpus.txt", "w", encoding="utf-8") as f:
    for doc in corpus_texts:
        if len(doc.split()) > 3:     # skip empty
            f.write(doc + "\n")
```

```python
import logging
logging.basicConfig(
    format="%(asctime)s : %(levelname)s : %(message)s",
    level=logging.INFO
)

logging.info("Logging works!")
```

```python
from gensim.models.word2vec import LineSentence

sentences = LineSentence("armenian_corpus.txt")

for s in sentences:
    print(s[:15])
    break
```

```python
from gensim.models import Word2Vec

model = Word2Vec(
    vector_size=300,
    window=5,
    sg=1,
    negative=15,
    min_count=5,
    workers=4,
    compute_loss=True
)

model.build_vocab(sentences)

prev = 0
for epoch in range(20):
    model.train(sentences, total_examples=model.corpus_count, epochs=1)
    loss = model.get_latest_training_loss()
    print(f"Epoch {epoch+1}/20 - Loss: {loss - prev}")
    prev = loss
```

```python
from gensim.models.word2vec import LineSentence

sentences = list(LineSentence("armenian_corpus.txt"))
print("Number of sentences:", len(sentences))
print(sentences[0][:20])
```

```python
from gensim.models import Word2Vec

model = Word2Vec(
    vector_size=300,
    window=5,
    sg=1,
    negative=15,
    min_count=5,
    workers=4,
    compute_loss=True
)

model.build_vocab(sentences)
```

```python
prev = 0
EPOCHS = 20

for epoch in range(EPOCHS):
    model.train(sentences, total_examples=model.corpus_count, epochs=1)
    loss = model.get_latest_training_loss()
    print(f"Epoch {epoch+1}/{EPOCHS} - Loss: {loss - prev}")
    prev = loss
```

```python
model.save("armenian_word2vec.model")
print("Saved!")
```

```python
accuracy, sections = model.wv.evaluate_word_analogies("word_analogy_task_hy.txt")

print("Accuracy:", accuracy)
```
