```python
"""
Part-of-Speech (POS) Tagging using HMM with Library-based Viterbi Algorithm

This version uses existing libraries to simplify the implementation:
- Uses NLTK's HiddenMarkovModelTagger which includes Viterbi algorithm
- Much shorter code
- Still maintains educational value with detailed comments

TRADE-OFFS:
===========
Pros:
+ Much simpler and shorter code
+ Well-tested, optimized implementations
+ Easier to maintain
+ Can handle larger datasets efficiently

Cons:
- Less visibility into the algorithm's inner workings
- Need to install external libraries
- Less educational for understanding the math

For production use, libraries are recommended.
For learning, the manual implementation (previous file) is better.
"""

import nltk
from collections import defaultdict


class LibraryHMMPOSTagger:
    """
    POS Tagger using NLTK's Hidden Markov Model with built-in Viterbi algorithm.

    This wraps NLTK's HiddenMarkovModelTagger for easier use.
    """

    def __init__(self):
        """
        Initialize the tagger.

        The HMM tagger will be created after training.
        """
        self.hmm_tagger = None
        self.trained = False


    def train(self, tagged_sentences):
        """
        Train the HMM tagger on tagged sentences.

        Args:
            tagged_sentences: List of sentences, where each sentence is a
                            list of (word, tag) tuples
                            Example: [[("The", "DET"), ("dog", "NOUN")], ...]

        How it works:
        -------------
        NLTK's HiddenMarkovModelTagger.train() automatically:
        1. Counts all transitions (tag to tag)
        2. Counts all emissions (tag to word)
        3. Computes transition probabilities P(tag_j | tag_i)
        4. Computes emission probabilities P(word | tag)
        5. Creates a Viterbi decoder for inference

        All the math we implemented manually is done by the library!
        """

        print("Starting training with NLTK HMM...")
        print(f"Training on {len(tagged_sentences)} sentences...")

        # Train the HMM tagger
        # This single line does all the work we did manually:
        # - Counting transitions and emissions
        # - Computing probabilities
        # - Setting up the Viterbi decoder
        self.hmm_tagger = nltk.HiddenMarkovModelTagger.train(tagged_sentences)

        self.trained = True

        # Extract some statistics for display
        # The tagger stores probabilities internally
        tags = set()
        vocab = set()

        for sentence in tagged_sentences:
            for word, tag in sentence:
                tags.add(tag)
                vocab.add(word)

        print(f"Training complete!")
        print(f"Learned {len(tags)} POS tags: {sorted(tags)}")
        print(f"Vocabulary size: {len(vocab)} words")


    def tag_sentence(self, sentence):
        """
        Tag a sentence using the trained HMM and Viterbi algorithm.

        Args:
            sentence: List of words (strings) or a single string

        Returns:
            List of (word, tag) tuples

        How it works:
        -------------
        The .tag() method internally:
        1. Runs the Viterbi algorithm to find the most likely tag sequence
        2. Returns the words paired with their predicted tags

        The Viterbi algorithm (built into the library):
        - Uses dynamic programming to efficiently find the best path
        - Considers both transition and emission probabilities
        - Finds the globally optimal tag sequence
        """

        if not self.trained:
            raise ValueError("Model must be trained before tagging!")

        # Handle string input
        if isinstance(sentence, str):
            sentence = sentence.split()

        # Tag the sentence using the trained HMM
        # This internally runs the Viterbi algorithm
        tagged = self.hmm_tagger.tag(sentence)

        return tagged


# ============================================================================
# EXAMPLE USAGE
# ============================================================================

def main():
    """
    Demonstrate the library-based HMM POS tagger.
    """

    print("=" * 70)
    print("HMM POS TAGGER USING NLTK LIBRARY")
    print("=" * 70)
    print()

    # TRAINING DATA
    # =============
    # Same training data as before

    training_data = [
        [("The", "DET"), ("dog", "NOUN"), ("barks", "VERB")],
        [("The", "DET"), ("cat", "NOUN"), ("meows", "VERB")],
        [("A", "DET"), ("dog", "NOUN"), ("runs", "VERB")],
        [("The", "DET"), ("dog", "NOUN"), ("runs", "VERB")],
        [("A", "DET"), ("cat", "NOUN"), ("sleeps", "VERB")],
        [("The", "DET"), ("big", "ADJ"), ("dog", "NOUN"), ("barks", "VERB")],
        [("A", "DET"), ("small", "ADJ"), ("cat", "NOUN"), ("meows", "VERB")],
        [("The", "DET"), ("dog", "NOUN"), ("sleeps", "VERB")],
        [("The", "DET"), ("cat", "NOUN"), ("runs", "VERB")],
        [("A", "DET"), ("big", "ADJ"), ("dog", "NOUN"), ("sleeps", "VERB")],
        [("The", "DET"), ("small", "ADJ"), ("cat", "NOUN"), ("runs", "VERB")],
        [("Dogs", "NOUN"), ("bark", "VERB")],
        [("Cats", "NOUN"), ("meow", "VERB")],
        [("The", "DET"), ("dog", "NOUN"), ("in", "ADP"), ("the", "DET"),
         ("house", "NOUN"), ("barks", "VERB")],
        [("A", "DET"), ("cat", "NOUN"), ("on", "ADP"), ("the", "DET"),
         ("mat", "NOUN"), ("sleeps", "VERB")],
    ]

    # CREATE AND TRAIN THE TAGGER
    # ===========================

    tagger = LibraryHMMPOSTagger()
    tagger.train(training_data)

    print()
    print("=" * 70)
    print("TAGGING NEW SENTENCES")
    print("=" * 70)
    print()

    # TEST SENTENCES
    # ==============

    test_sentences = [
        ["The", "dog", "barks"],
        ["A", "cat", "sleeps"],
        ["The", "big", "cat", "runs"],
        ["Dogs", "bark"],
        ["The", "cat", "in", "the", "house", "meows"],
    ]

    # Tag each test sentence
    for sentence in test_sentences:
        print(f"Sentence: {' '.join(sentence)}")

        # Get tagged result using library's Viterbi
        tagged = tagger.tag_sentence(sentence)

        # Print result
        print("Tagged:  ", end="")
        for word, tag in tagged:
            print(f"{word}/{tag} ", end="")
        print()
        print()

if __name__ == "__main__":
    main()
```
