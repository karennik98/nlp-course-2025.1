```python
text = """Natural Language Processing (NLP) is a subfield of linguistics, computer science, and artificial
intelligence concerned with the interactions between computers and human language. It's used to
analyze text, allowing machines to understand, interpret, and manipulate human language. NLP has
many real-world applications, including machine translation, sentiment analysis, and chatbots."""

tokens_step1 = text.split()

print("tokens:")
print(tokens_step1)
```

```python
tokens_lower = [t.lower() for t in tokens_step1]

print("lowercase:")
print(tokens_lower)
```

```python
def remove_punct(token: str) -> str:
    cleaned = "".join(ch for ch in token if ch.isalnum() or ch == "'")
    cleaned = cleaned.strip("'")
    return cleaned

tokens_step2_3 = [remove_punct(t) for t in tokens_lower]
tokens_step2_3 = [t for t in tokens_step2_3 if t]  # remove empty tokens

print("lowercase + no punctuation:")
print(tokens_step2_3)
```

```python
stop_words = {"the", "and", "is", "are", "a", "an", "in", "on", "at", "for", "to", "of"}

tokens_step4 = [t for t in tokens_step2_3 if t not in stop_words]

print("stopwords removed:")
print(tokens_step4)
```

```python
suffixes = ["ing", "ed", "ly", "es", "s", "ment", "tion", "ions", "er", "ers"]

def simple_stem(word: str) -> str:
    for suf in sorted(suffixes, key=len, reverse=True):
        if word.endswith(suf) and len(word) > len(suf) + 2:
            return word[:-len(suf)]
    return word

tokens_step5 = [simple_stem(t) for t in tokens_step4]

print("stemming:")
print(tokens_step5)
```
