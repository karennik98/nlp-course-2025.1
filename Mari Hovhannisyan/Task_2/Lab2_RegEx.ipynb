Email Addresses

```python
text = """Contact us at support@example.com or sales@company.org for assistance.
For personal inquiries, email john.doe123@university.edu."""
import re

pattern = r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}"
emails = re.findall(pattern, text)

for e in emails:
    print(e)
```

Problem 2: Validate Phone Numbers

```python
text = "Valid: 123-456-7890, 987-654-3210\nInvalid: 12-345-67890, 1234567890, 123-45-6789"
pattern = r"\b\d{3}-\d{3}-\d{4}\b"
valid_numbers = re.findall(pattern, text)

print("Valid phone numbers:")
for num in valid_numbers:
    print(num)
```

Extract Dates

```python
text = "Important dates: 25/12/2023, 01-01-2024, 31/05/2023, and 15-10-2024."
pattern = r"\b\d{2}[/-]\d{2}[/-]\d{4}\b"
dates = re.findall(pattern, text)

for d in dates:
    print(d)
```

Repeated Words

```python
text = "The the quick brown fox jumps over the the lazy dog."
pattern = r"\b([A-Za-z]+)\s+\1\b"
repeated = re.findall(pattern, text, flags=re.IGNORECASE)

print("Repeated words:")
for w in repeated:
    print(f"{w.lower()} {w.lower()}")
```

Extract Hashtags

```python
text = "Check out our new products: #Sale2024, #NewArrival, and #Discounts!"
pattern = r"#\w+"
hashtags = re.findall(pattern, text)

for h in hashtags:
    print(h)
```

Validate passwords

```python
passwords = ["Password123", "Secure456", "weak", "password", "Password"]
pattern = r"^(?=.*[A-Z])(?=.*[a-z])(?=.*\d).{8,}$"

valid_passwords = [p for p in passwords if re.match(pattern, p)]

print("Valid passwords:")
for p in valid_passwords:
    print(p)
```

Extract URLs

```python
text = "Visit our website at https://www.example.com or check out http://blog.example.org for updates."
pattern = r"https?://[^\s]+"
urls = re.findall(pattern, text)

urls = [u.rstrip(".,!?:;") for u in urls]

for u in urls:
    print(u)
```

Multiple Spaces

```python
text = "This   text   has    multiple   spaces."
result = re.sub(r"\s+", " ", text).strip()
print(result)
```

Extract Quoted Text


```python
text = 'He said, "Hello, world!" and she replied, "Hi there!"'
pattern = r'"(.*?)"'
quoted = re.findall(pattern, text)

for q in quoted:
    print(q)
```

Validate IP Addresses
Problem 10: Validate IP Addresses

```python
text = "Valid: 192.168.1.1, 10.0.0.255\nInvalid: 256.1.2.3, 192.168.01.1, 192.168.1"
octet = r"(?:25[0-5]|2[0-4]\d|1\d\d|[1-9]?\d)"
pattern = rf"\b{octet}\.{octet}\.{octet}\.{octet}\b"

valid_ips = re.findall(pattern, text)

print("Valid IP addresses:")
for ip in valid_ips:
    print(ip)
```
